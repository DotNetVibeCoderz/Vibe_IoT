# Which protocol for which job?

| You need to… | Use | IoTCom.Net package |
|---|---|---|
| Read/write a PLC, meter, VFD or I/O module | Modbus TCP/RTU | `Protocols.Modbus` |
| Ship telemetry to the cloud or many consumers | MQTT (+ SenML payloads) | `Adapters.Mqtt`, `Serialization.SenML` |
| Read position/time from a GPS or marine instruments | NMEA 0183 | `Protocols.Nmea` |
| Drive stage or architectural lighting | Art-Net or sACN | `Protocols.Dmx` |
| Collect battery sensors kilometres away | LoRaWAN (Semtech UDP gateways) | `Protocols.LoRaWan` |
| Read electricity meters / building sub-meters | DLMS/COSEM / M-Bus | `Protocols.Dlms`, `Protocols.MBus` |
| Talk to a microcontroller over UART/USB with your own messages | COBS or SLIP + CRC | `Framing` |
| Debug a link byte by byte | Traffic tap, `iotcom modbus decode`, Gallery workbench | `Core`, CLI |

**Rules of thumb.** Polling one device on a LAN → Modbus TCP. Fan-out to many consumers or over the internet → MQTT with TLS. Sensor payloads crossing vendor boundaries → SenML. Never expose Modbus to the internet: put a gateway in front.
